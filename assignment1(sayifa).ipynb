```python
# Recharge validity calculator

recharge = input("Enter recharge date (yyyy-mm-dd): ")
today = input("Enter today's date (yyyy-mm-dd): ")

# Convert recharge date into total days
y, m, d = recharge.split("-")
recharge_days = int(y) * 365 + int(m) * 30 + int(d)

# Convert today's date into total days
y, m, d = today.split("-")
today_days = int(y) * 365 + int(m) * 30 + int(d)

# Calculate expiry date
expiry_days = recharge_days + 28

# Compare
if today_days > expiry_days:
    print("Plan expired. Recharge again.")
else:
    print("Your plan is still valid.")
```

```python

#Dictionary – Student Marks Summary
students = {'s': [], 'm': []}

n = 3

while n != 0:

    st = input("Enter the name of the student: ")
    mark = int(input("Enter the mark: "))

    students['s'].append(st)
    students['m'].append(mark)

    n -= 1

total = 0

for mark in students['m']:
    total = total + mark

average = total / 3

print("Students =", students['s'])
print("Marks =", students['m'])
print("Average mark of student =", average)
```

```python
#Dictionary – Word Frequency
sentence = input("Enter a sentence: ")
words = sentence.split()
frequency = {}
for word in words:

    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1
for word in frequency:
    print(word, ":", frequency[word])
```

```python
#functions – Check Prime
def prime(n):
  count=0
  for i in range(2,n):
    if n%i==0:
      count==1
      break
  if count==1:
      print(n," is not prime")

  else:
      print(n," is prime ")

n=int(input("enter the number to check prime or not"))
prime(n)
```

```python
#Vowel Counter
nv,nc=0,0
vowels=['a','e','i','o','u','A','E','I','O','U']
st=input("enter the string  ")
for i in st:
  if i in vowels:
    nv+=1
  else:
    nc+=1
print("no.of vowels= ",nv," no.of consonents = ",nc)


```
